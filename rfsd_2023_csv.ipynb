```python
import pyarrow.parquet as pq

# Открываем файл за 2023 год
source = pq.ParquetFile("part-0.parquet")

# Выбираем строки баланса и отчёта о финансовых результатах
columns = [
    "line_1100", "line_1150", "line_1200", "line_1210", "line_1230",
    "line_1250", "line_1300", "line_1370", "line_1400", "line_1500",
    "line_1520", "line_1600", "line_1700", "line_2110", "line_2120",
    "line_2200", "line_2300", "line_2340", "line_2350", "line_2400",
    ]

# Берём первые 1 000 записей
batches = source.iter_batches(batch_size=1000, columns=columns)
df = next(batches).to_pandas()

# Сохраняем таблицу без индекса и выводим что получилось
df.to_csv("rfsd_2023.csv", index=False)
print(df.shape)
```
